# Plotting port outputs

The ``toolboxes.Plotting.plot_port`` module reads the authoritative terminal quantities stored by gprMax. It discovers voltage-source ports, rational-network ports, transmission lines, magnetic-frill ports, and ports in subgrids. Depending on the port type, the output can contain complex $S_{11}$, input impedance and admittance, validity diagnostics, and available voltage and current histories and spectra. The plotting tool never recalculates these quantities from raw receiver fields.

A voltage source owns its port result directly, as do transmission-line and magnetic-frill sources. Useful command-line options include:

* ``--list-ports`` lists all terminal-output groups in the file.
* ``--port`` selects a port by ID or complete HDF5 path and may be repeated. All ports are plotted by default.
* ``--fmin`` and ``--fmax`` limit the plotted frequency range.
* ``--parameters-only`` omits the available time histories and spectra.
* ``--validity`` adds the source-band, mesh, and correction diagnostics.
* ``--save`` writes uniquely named figures; ``--format`` selects PNG, PDF, or SVG.

For example, to save the wire-dipole port figures:

    python -m toolboxes.Plotting.plot_port antenna_wire_dipole_fs.h5 --port feed --validity --save

The compact model below is generated automatically for the notebook. Replace ``filename`` with any gprMax HDF5 file containing terminal output.
