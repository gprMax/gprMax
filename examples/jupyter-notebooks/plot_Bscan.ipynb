# Plotting a B-scan

The ``toolboxes.Plotting.plot_Bscan`` module plots a B-scan (a collection of A-scans) from a merged gprMax HDF5 output file. It takes the output filename and a field or current component, i.e. ``Ex``, ``Ey``, ``Ez``, ``Hx``, ``Hy``, ``Hz``, ``Ix``, ``Iy``, or ``Iz``, provided that component was stored by the model.

For example (to use the module outside this notebook) to plot the B-scan of the ``Ez`` component of an output:

    python -m toolboxes.Plotting.plot_Bscan cylinder_Bscan_2D_merged.h5 Ez

You can use the following code to experiment (in this notebook) with plotting different field/current components.

```python
%matplotlib inline
from notebook_helpers import ensure_bscan
from toolboxes.Plotting.plot_Bscan import mpl_plot
from toolboxes.Utilities.outputfiles_merge import get_output_data

filename = ensure_bscan()
rxnumber = 1
rxcomponent = 'Ez'
outputdata, dt = get_output_data(filename, rxnumber, rxcomponent)
plt = mpl_plot(filename, outputdata, dt, rxnumber, rxcomponent)

# Change from the default 'seismic' colormap
#plt.set_cmap('gray')
```
