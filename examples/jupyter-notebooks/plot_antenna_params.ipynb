# Plotting antenna parameters

The ``toolboxes.Plotting.plot_antenna_params`` module plots terminal data stored by ``#rx_port``/``RxPort``. It discovers voltage-source, transmission-line, and magnetic-frill terminal outputs and plots the quantities that each source type makes available. These include complex $S_{11}$, input impedance and admittance, plus available voltage and current histories and spectra.

The model must contain an ``RxPort`` at exactly the same position as a compatible source. The module takes the HDF5 output filename and these useful optional arguments:

* ``--list-ports`` lists all terminal-output groups in the file.
* ``--port`` selects a port by ID or complete HDF5 path; it is optional when the file contains one port.
* ``--fmin`` and ``--fmax`` limit the plotted frequency range.
* ``--params-only`` omits the available time histories and spectra.

For example (to use the module outside this notebook) to plot the antenna parameters from the wire dipole antenna:

    python -m toolboxes.Plotting.plot_antenna_params antenna_wire_dipole_fs.h5 --port feed

The compact model below is generated automatically for the notebook. Replace ``filename`` with any gprMax HDF5 file containing terminal output.

```python
%matplotlib inline
from notebook_helpers import ensure_port_demo
from toolboxes.Plotting.plot_antenna_params import mpl_plot_port, read_port_params

filename = ensure_port_demo()
portparams = read_port_params(filename, port_id='feed')
plt = mpl_plot_port(filename, **portparams)
```
